## 2. Explore Category Schema

Let's examine the 8 categories and their associated keywords.

```python
# Display category information
category_info = []
for cat_name, cat_data in CATEGORIES.items():
    category_info.append({
        'Category': cat_name,
        'Tag': cat_data['tag'],
        'Description': cat_data['description'],
        'Keywords': len(cat_data['keywords']),
        'Sample Keywords': ', '.join(cat_data['keywords'][:5])
    })

df_categories = pd.DataFrame(category_info)
display(df_categories)

# Visualize keyword counts
plt.figure(figsize=(10, 5))
plt.barh(df_categories['Category'], df_categories['Keywords'], color='skyblue')
plt.xlabel('Number of Keywords')
plt.title('Keywords per Category')
plt.tight_layout()
plt.show()
```

## 3. Initialize Classifier and Test Classification

Initialize the PageClassifier and test on sample content.

```python
# Initialize classifier
classifier = PageClassifier()

print("✓ Classifier initialized")
print(f"Model: {classifier.model_name}")
print(f"Categories: {len(classifier.categories)}")

# Test with sample texts
test_samples = [
    ("https://sdit.ac.in/about/", "Shree Devi Institute of Technology was established in 2006. Our vision is to provide quality education and create skilled engineers. The principal and management are committed to excellence."),
    ("https://sdit.ac.in/departments/cse/", "The Computer Science department offers BE in CSE and IT. Our faculty members are experienced in programming, algorithms, and software engineering."),
    ("https://sdit.ac.in/admissions/", "Admission process for 2024. Eligibility criteria: 45% in PUC. Application form available online. Fee structure and important dates."),
    ("https://sdit.ac.in/placements/", "Our placement cell has achieved 95% placement rate. Top recruiters include Infosys, TCS, and IBM. Training programs prepare students for interviews."),
]

results = []
for url, text in test_samples:
    category, tag, confidence = classifier.classify(text, url)
    results.append({
        'URL': url.split('/')[-2],
        'Category': category,
        'Tag': tag,
        'Confidence': f"{confidence:.2f}"
    })

df_results = pd.DataFrame(results)
display(df_results)
```
